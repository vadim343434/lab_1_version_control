# **Завдання 2**
Намалювати графік функції   $y = sin(x)$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-7.3, 8.1, 400)
y = np.sin(x)

plt.plot(x, y)

plt.xlabel("x")
plt.ylabel("y = sin(x)")
plt.title("Графік функції y = sin(x)")
plt.grid(True)

plt.show()
```

# **Завдання 3**
Дано 2 точки A і B з відповідними координатами ($x_a, y_a$) та ($x_b, y_b$). Точка C з координатами ($x_c, y_c$).  
- Намалювати графік, на якому відображатимуться точки A, B, C.
- Точка А повинна бути червоним кольором.
- Точка С повина мати квадратну форму маркера.
- Знайти відстань між точками A і B за формулою евклідової відстані:

$d_{EUCLID}(O_1,O_2) = \sqrt{(x_1-x_2)^2+(y_1-y_2)^2}$

```python
import math
import matplotlib.pyplot as plt

# Координати точок
xa, ya = 4.7, -2.4
xb, yb = -3.7, 5.1
xc, yc = 1.8, 6.5

# Обчислення евклідової відстані між A та B
distance = math.sqrt((xa - xb)**2 + (ya - yb)**2)

print("Відстань між точками A і B = %.2f" % distance)

# Побудова точок
plt.scatter(xa, ya, color="red", marker="o", s=80, label="A")
plt.scatter(xb, yb, marker="o", s=80, label="B")
plt.scatter(xc, yc, marker="s", s=80, label="C")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Графік точок A, B та C")
plt.grid(True)
plt.legend()

plt.show()
```
